Q3. Highest sales date in France

Considering the given Sales data, on which date has the highest sales occured in France?

Dataset: https://drive.google.com/file/d/1UViHcT4lKDsTfc97gBqIvds25uHe4b5e/view?usp=drive_link

Note:

*  'sales' is calculated by multiplying Quantity and Price
*  The date is formatted as YYYY-MM-DD

Choose the correct answer from below:

2011-02-31

2011-09-30

2010-01-27

2010-08-09

```python
import pandas as pd
file_id = '1UViHcT4lKDsTfc97gBqIvds25uHe4b5e'
df = pd.read_csv(f'https://drive.google.com/uc?id={file_id}')

df.head()
```

```python
df = df[df['Country']=='France']

df['total'] = df['Quantity']*df['Price']
df['date'] = pd.to_datetime(df['InvoiceDate']).dt.strftime('%Y-%m-%d')
df.groupby(by='date')['total'].sum().idxmax()
```
