### Cosine Similarity

Recall
- how Euclidean distance is the hypothesis distance between two points
- and how Manhattan distance is the distance covered along the axes

- We also know, how Euclidean Distance fails due to Curse of Dimensionality.

- Notice how both Manhatthan and Euclidean covers the magnitude of the distance
#### What if we now look into the angular Distance between two points?

- Lets understand this with an example
- Suppose we have two points $x_1$ and $x_2$
    - which are very close to each other,
    - and are in the same vector direction.
    - If we measure the Angular distance and discard

#### How do we measure these Angular distance ?
- Using the cosine function  (cosθ)
- Since if two points are similar, the Angular distance becomes 0, then cosθ = 1


If we have three points $x_1$,$x_2$,$x_3$ which have angles of $θ$ and $ϕ$
- Cosθ and cosϕ becomes are angular distance

#### How can we calculate Cosθ for two d-dimensional features ?
Lets say we want to find how similar $x_1$ and $x_2$ are.
- Remember since $x_1$ and $x_2$ are just d-dimensional vectors
- Cosθ is just the dot product of these two vectors divided by the Norm of each vector.
    - $Cosθ_{12} = \frac{x_1.x_2}{||x_1||.||x_2||}$

#### If d very large, wont calculation of  norm of Vector be a time consuming task ?
- Recall how data is normalized, and norm of Normalized data = 1,
- Hence $Cosθ_{12} = x_1.x_2$

Before we dive into LSH lets first understand how hashing works by taking an example

- Suppose we have [1,12,11,16,18]
as our data and we want to put this data into a hash table
#### How can we perform Hashing ?
- The most simplest function to use is Mod,
    - since Mod is finding remainders
    - And we know how remainder, cycles between 0 and n-1 for a number n  
    - Eg. if we do mod of 3
        - 1 mod 3  = 1
        - 2 mod 3 = 2
        - 3 mod 3 = 0
        - 4 mod 3 = 1
        - 5 mod 3 = 2
        - 6 mod 3 = 0
    - Notice the repeated cycle between 0 to 3-1 = 2.
#### What does Hash table consists of?
- Keys and Values

Now for the example:
- we use mod 10 to fill the hash table,
- 1 mod 10 = 1, hence the Key 1 will have value of 1

- 12 mod 10 = 2,hence the Key 2 will have value of 12

- 11 mod 10 = 1,hence the Key 1 will have value of [1,11]

- 16 mod 10 = 6,hence the Key 6 will have value of 16

- 18 mod 10 = 8,hence the Key 8 will have value of 18


Now if we have a query point of $q=11$,
- since 11 mod 10 = 1, it will directly go to key 1 and return 11

#### what if we have a query point of $q=21$?
- since 21 mod 10 = 1, it will directly go to key 1 and returns no element found
- since it does not contain any value = 21  


if we notice, Hashing just groups elements together e.g 1 and 11, and then searches within that group only

With the same grouping approach in mind, We have two features $f_1$,$f_2$
- And we create a first random plane $π_1$ which passes through origin
#### What will be the equation of $π_1$?
- $w^Tx = 0 $

Now we create another hyperplane $π_2$ which passes through origin
- Notice how these two hyperplanes divides the datapoints into 4 regions:
    - Where $Π_1$ +ve and $Π_2$ +ve
    - Where $Π_1$ +ve and $Π_2$ -ve
    - Where $Π_1$ -ve and $Π_2$ +ve
    - Where $Π_1$ -ve and $Π_2$ -ve  

Now all these regions becomes our key for the hash table, and the datapoints inside them becomes our values
- For a new query point $x_q$, we first find which region it belongs to, and then find the K- Nearest Neighbor
- This reduces time complexity, since we are looking into only a samll Region instead of the entire datapoint.

### KNN for Regression

- Remember how Logisitc Regression can be used only for Classification Problems

#### Can we also use KNN for Regression problem ?
Lets assume we have some datapoints whose $y_i$  are real number (since its a regression problem)
- And we have a query point $x_q$
- Now if we take K = 5
- We get 5 $Y_i$ of nearest neighbors of $x_q$
